## Loan Default and Total Loan Amount Prediction

This project focuses on building machine learning models to predict **loan default risk** and **total loan amount eligibility** based on applicant financial and demographic data.

The objective is twofold:

1. **Predict whether a borrower will default on a loan** (classification problem).
2. **Predict the total loan amount to be approved** for a borrower (regression problem).

By analyzing historical loan data and key features such as income, credit history, employment status, and loan terms, this project aims to support **data-driven decision-making in financial lending**. The models developed can help financial institutions reduce risk, improve approval accuracy, and optimize loan allocation.
